Series

```python
import pandas as pd
a=[10,20,30]
s= pd.Series(a,name='marks')
print(s)#creating a series using ng list
```

```python
r= range(100)
a= list(r)
s= pd.Series(a)
print(s)#using range and list
```

```python
import pandas as pd
import numpy as np
a=[10,20,30]
i=["a","b","c"]
data= np.array(a)
print(data)#array and data type printing
print(type(data))

d= pd.Series(a, index = i) #index changing in series

print(d)
```

NAN values(not a number)

```python
import pandas as pd
import numpy as np

a= (4,8,np.nan)
b=["car","bat",np.nan]
s= pd.Series(a)
z= pd.Series(b)
print(s)
print(z)
```
